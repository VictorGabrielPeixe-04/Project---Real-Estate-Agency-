Não use bibliotecas externas (como Weka ou DeepNetts) num primeiro momento. Implemente uma classe RegressaoEspacial que:

```python
import numpy as np
```

```python
planet = np.array(['Alpha-1', 'Beta-1', 'Gamma-3', 'Delta-4', 'Epsilon-5'])

distance_x = np.array([0.8, 1.2, 1.5, 0.5, 2.0])

indice_hab_y = np.array([85,  70,  60, 40, 30])
```

```python
#CALCULANDO A MÉDIA DE X E Y
mean_x = np.mean(distance_x)

mean_y = np.mean(indice_hab_y)
```

```python
#Média
print(f"Média distância x: {mean_x} ")

#Média Y
print(f"Média índice de habitalidade: {mean_y}")
```

```python
#Calculando os coeficientes
b1 = np.sum((distance_x - mean_x) * (indice_hab_y - mean_y)) / np.sum((distance_x - mean_x) ** 2)

b0 = mean_y - b1 * mean_x
```

```python
#Resultado inclinação
print(f"Resultado inclinação: {b1}")

#Resultado Intercepto
print(f"Intercepto: {b0}")
```

```python
#Calculando método prever
def foresees(distance_x_, b1, b0):
    return b0 + (b1 * distance_x_)
```

```python
#Previsão
forecast =  foresees(1.1, b0, b1)
print(f"Previsão Z-java : {forecast:.2f}")
```

```python
#Previsão para todos os planetas
calc_forecast_all = b0 + (b1 * distance_x)
```

```python
print(calc_forecast_all)
```

```python
mse = np.mean((indice_hab_y - calc_forecast_all) ** 2)
```

```python
print(f"MSE: {mse:.2f}")
```
